```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Chargement des données
jeu = load_diabetes()
X = jeu.data
y = jeu.target

# Séparation des données
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Tester deux configurations
for valeur in [True, False]:

    modele = LinearRegression(fit_intercept=valeur)
    modele.fit(X_train, y_train)

    predictions = modele.predict(X_test)

    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    print("fit_intercept =", valeur)
    print("RMSE :", rmse)
    print("R² :", r2)
    print("-------------------")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Séparation train/test : régression
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=42
)

# Séparation train/test : classification
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_cls, y_cls, test_size=0.20, random_state=42, stratify=y_cls
)

# Colonnes numériques et catégorielles
num_cols = Xc_train.select_dtypes(include=np.number).columns
cat_cols = Xc_train.select_dtypes(exclude=np.number).columns

# Prétraitement numérique
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Prétraitement catégoriel
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_cls = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

print("Train régression :", Xr_train.shape)
print("Test régression :", Xr_test.shape)
print("Train classification :", Xc_train.shape)
print("Test classification :", Xc_test.shape)
```

```python
# Exclure duration pour une prédiction réaliste avant l'appel
if "duration" in Xc_train.columns:
    Xc_train = Xc_train.drop(columns=["duration"])
    Xc_test = Xc_test.drop(columns=["duration"])

num_cols = Xc_train.select_dtypes(include=np.number).columns
cat_cols = Xc_train.select_dtypes(exclude=np.number).columns

preprocessor_cls = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

```python
# Vérifier les noms des colonnes
print(df_retail.columns.tolist())

# Nettoyage
df_retail.columns = df_retail.columns.str.strip()

df_retail["InvoiceDate"] = pd.to_datetime(
    df_retail["InvoiceDate"], errors="coerce"
)

df_retail = df_retail.dropna(
    subset=["CustomerID", "InvoiceDate"]
).drop_duplicates()

# Supprimer les annulations et les valeurs non valides
df_retail = df_retail[
    (~df_retail["InvoiceNo"].astype(str).str.startswith("C")) &
    (df_retail["Quantity"] > 0) &
    (df_retail["UnitPrice"] > 0)
].copy()

df_retail["Montant"] = (
    df_retail["Quantity"] * df_retail["UnitPrice"]
)

# Date de référence
date_ref = df_retail["InvoiceDate"].max() + pd.Timedelta(days=1)

# Création des indicateurs RFM
rfm = df_retail.groupby("CustomerID").agg(
    Recence=("InvoiceDate", lambda x: (date_ref - x.max()).days),
    Frequence=("InvoiceNo", "nunique"),
    Montant=("Montant", "sum")
)

print("Dimensions RFM :", rfm.shape)
display(rfm.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Garder les clients ayant des valeurs positives
rfm = rfm[(rfm > 0).all(axis=1)].copy()

# Transformation logarithmique pour limiter l'effet des grandes valeurs
X_rfm_log = np.log1p(rfm)

# Standardisation
scaler_rfm = StandardScaler()
X_rfm = scaler_rfm.fit_transform(X_rfm_log)

print("Données finales :", X_rfm.shape)
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

modeles_reg = {
    "Baseline": DummyRegressor(strategy="mean"),
    "Random Forest": RandomForestRegressor(
        n_estimators=200, random_state=42
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}

resultats_reg = []

for nom, modele in modeles_reg.items():
    pipeline = make_pipeline(
        SimpleImputer(strategy="median"),
        modele
    )

    pipeline.fit(Xr_train, yr_train)
    predictions = pipeline.predict(Xr_test)

    resultats_reg.append({
        "Modèle": nom,
        "MAE": np.mean(np.abs(yr_test - predictions)),
        "RMSE": np.sqrt(np.mean((yr_test - predictions) ** 2)),
        "R2": pipeline.score(Xr_test, yr_test)
    })

df_resultats_reg = pd.DataFrame(resultats_reg)
display(df_resultats_reg.round(3))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score
)

modeles_cls = {
    "Baseline": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "Random Forest": RandomForestClassifier(
        n_estimators=200, random_state=42,
        class_weight="balanced"
    )
}

resultats_cls = []

for nom, modele in modeles_cls.items():
    pipeline = Pipeline([
        ("preprocessing", preprocessor_cls),
        ("model", modele)
    ])

    pipeline.fit(Xc_train, yc_train)
    pred = pipeline.predict(Xc_test)

    ligne = {
        "Modèle": nom,
        "Accuracy": accuracy_score(yc_test, pred),
        "Precision": precision_score(
            yc_test, pred, pos_label="yes", zero_division=0
        ),
        "Recall": recall_score(
            yc_test, pred, pos_label="yes", zero_division=0
        ),
        "F1": f1_score(
            yc_test, pred, pos_label="yes", zero_division=0
        )
    }

    if hasattr(pipeline, "predict_proba"):
        probabilites = pipeline.predict_proba(Xc_test)
        classes = list(pipeline.classes_)
        index_yes = classes.index("yes")
        ligne["ROC_AUC"] = roc_auc_score(
            (yc_test == "yes").astype(int),
            probabilites[:, index_yes]
        )

    resultats_cls.append(ligne)

df_resultats_cls = pd.DataFrame(resultats_cls)
display(df_resultats_cls.round(3))
```

```python
!pip install -q gymnasium

import gymnasium as gym
import numpy as np

env = gym.make("Taxi-v4")

def entrainer_q_learning(env, episodes=5000, alpha=0.1,
                         gamma=0.99, epsilon=0.1, seed=42):

    Q = np.zeros((env.observation_space.n,
                  env.action_space.n))

    rng = np.random.default_rng(seed)

    for episode in range(episodes):
        state, info = env.reset(seed=seed + episode)
        termine = False

        while not termine:
            if rng.random() < epsilon:
                action = env.action_space.sample()
            else:
                action = np.argmax(Q[state])

            next_state, reward, terminated, truncated, info = env.step(action)
            termine = terminated or truncated

            cible = reward
            if not termine:
                cible += gamma * np.max(Q[next_state])

            Q[state, action] += alpha * (
                cible - Q[state, action]
            )

            state = next_state

    return Q

Q = entrainer_q_learning(env)
print("Apprentissage Q-learning terminé.")
```

## 27. Q-learning

**Question ciblée (Partie B) :** Comment calculer une cible de Bellman dans Taxi et pourquoi le max intervient-il ?

**À compléter par le binôme :** présentation · schéma original · deux cas d’usage · analyse de la sortie et variation d’un hyperparamètre.

```python
# Q-learning manuel sur Taxi-v3 : exploration epsilon-greedy
import gymnasium as gym
env=gym.make('Taxi-v3');Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(3000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.07,1-episode/2400)
    for pas in range(200):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .96*np.max(Q[suivant]))
        Q[etat,action]+=.2*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Q-learning Taxi — taille de Q:',Q.shape,'Valeurs du premier état:',Q[0].round(2));env.close()
```

## 28. SARSA

**Question ciblée (Partie B) :** Pourquoi SARSA met-il à jour Q avec l’action suivante réellement choisie ?

```python
# SARSA manuel sur Taxi-v4 (on-policy)
import gymnasium as gym
import numpy as np
env=gym.make('Taxi-v4');Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
def choisir_action(etat,epsilon):
    return int(generateur.integers(env.action_space.n)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(3000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.07,1-episode/2400);action=choisir_action(etat,epsilon)
    for pas in range(200):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir_action(suivant,epsilon)
        cible=recompense+(0 if termine else .96*Q[suivant,action_suiv])
        Q[etat,action]+=.2*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('SARSA Taxi — taille de Q:',Q.shape,'Valeurs du premier état:',Q[0].round(2));env.close()
```

```python
def entrainer_sarsa(env, episodes=5000, alpha=0.1,
                    gamma=0.99, epsilon=0.1, seed=42):

    Q = np.zeros((env.observation_space.n,
                  env.action_space.n))

    rng = np.random.default_rng(seed)

    def choisir_action(state):
        if rng.random() < epsilon:
            return env.action_space.sample()
        return np.argmax(Q[state])

    for episode in range(episodes):
        state, info = env.reset(seed=seed + episode)
        action = choisir_action(state)
        termine = False

        while not termine:
            next_state, reward, terminated, truncated, info = env.step(action)
            termine = terminated or truncated

            if not termine:
                next_action = choisir_action(next_state)
                cible = reward + gamma * Q[next_state, next_action]
            else:
                next_action = None
                cible = reward

            Q[state, action] += alpha * (
                cible - Q[state, action]
            )

            state = next_state
            if next_action is not None:
                action = next_action

    return Q

Q_sarsa = entrainer_sarsa(env)
print("Apprentissage SARSA terminé.")

env.close()
```
